**13. Nested if Statements**

```python
acc_balance = int(input("Enter Account Balance: "));
if acc_balance > 0:
  widrawal = int(input("Enter Withdrawal Amount: "));
  if widrawal <= acc_balance:
    acc_balance -= widrawal;
    print("Widrawl is Possible: ");
  else:
    print("Insufficient Balance");
else:
  print("No Balance is in account");
```
